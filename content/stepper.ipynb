```python
# Load the required libraries
import ipywidgets as widgets

play = widgets.Play(
    value=50,
    min=0,
    max=100,
    step=1,
    interval=250,
    description="Press play",
    disabled=False
)
slider = widgets.IntSlider()
widgets.jslink((play, 'value'), (slider, 'value'))
widgets.HBox([play, slider])
```
